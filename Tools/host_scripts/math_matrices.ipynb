```python
from  sympy import *
from sympy.abc import a, b, c, d
init_printing()
```

```python
# Angles to describe rotation of frame(!) in 3D
# Right-handed coordinate frame convention (Z is up)
theta, phi, ksi = symbols('theta, phi, ksi')
# print(type(theta))
# print(theta, phi, ksi)
```

```python
theta 
```

```python
# Rotation around  X (Roll)
Rroll = Matrix([[         1,             0,               0], 
                [          0,    cos(theta),     sin(-theta)], 
                [          0,    sin(theta),      cos(theta)]])

# Rotation around Y (Pitch)
Rpitch = Matrix([[ cos(phi),       0,         sin(phi)], 
                [         0,       1,                0], 
                [ sin(-phi),       0,         cos(phi)]])

# Rotation around Z (Yaw)
Ryaw = Matrix([[ cos(ksi),       sin(-ksi),        0], 
                [sin(ksi),        cos(ksi),        0], 
                [       0,               0,        1]])
```

```python
# R matrices are orthogonal: R = R^-1 = R^T
# simplify(Rroll.T * Rroll) #should be identical matrix
# simplify(Rpitch.T * Rpitch)  
# simplify(Ryaw.T * Ryaw) #should be identical matrix


simplify(Rpitch.inv() * Rpitch) #should be identical matrix
```

```python
# Directon Matrix (orthogonal)
C = Rroll * Rpitch * Ryaw
print(simplify(C.inv() * C) == simplify(C.T * C))
# simplify(C.T * C) == simplify(C, inverse=C)
```
